# 连续时间系统 (Continuous-Time Dynamics)

## 常见形式

$$\dot{x} =f( x,u)$$

- $f$
- $x$
- $u$

如果是机械系统, 如机器人系统:
$$x=\left[\begin{aligned}
                    q \\
                    v
                \end{aligned}\right]$$

- $q$
- $v$
