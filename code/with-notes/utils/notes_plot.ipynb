```julia
using LaTeXStrings
import PyPlot
x = repeat(LinRange(-2, 2, 20), inner=20)
y = repeat(LinRange(-2, 2, 20), outer=20)
u = -1 .* x + 0 .* y
v = 2  * x + 1 * y
```

```julia
PyPlot.rc("text", usetex = true)
PyPlot.rc("text.latex", preamble = "\\usepackage{{amsmath}}")
PyPlot.quiver(x, y, u, y)
PyPlot.title(L"$\dot x = \begin{bmatrix} -1 & 0 \\ 2 & 1 \end{bmatrix}x$")
PyPlot.savefig("b.pdf")
```
