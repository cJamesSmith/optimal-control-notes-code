```julia
using LinearAlgebra
using ForwardDiff
using PyPlot
using LaTeXStrings
using JupyterFormatter
enable_autoformat()
PyPlot.rc("text", usetex = true)
PyPlot.rc("font", family = "Times New Roman")
PyPlot.rc("text.latex", preamble = "\\usepackage{{amsmath}}")
```

```julia
function f(x)
    return x.^4+x.^3-x.^2-x;
end
dot_f = x -> ForwardDiff.derivative(x->f(x), x);
ddot_f = x -> ForwardDiff.derivative(x->ForwardDiff.derivative(x->f(x), x), x);
```

```julia
x = LinRange(-1.75, 1.25, 1000)
```

```julia
fig = figure(figsize = (4, 3))
plot(x, f(x))
```

```julia
function newton_step(x0)
    xn = x0 - ddot_f(x0) \ dot_f(x0)
end
```

```julia
fig = figure(figsize = (4, 3))
xguess = 0.0
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
xnew = newton_step(xguess[end])
xguess = [xguess; xnew]
fig = figure(figsize = (4, 3))
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
function regularized_newton_step(x0)
    beta = 1.0
    H = ddot_f(x0)
    while H <= 0  # isposdef https://docs.julialang.org/en/v1/stdlib/LinearAlgebra/#LinearAlgebra.isposdef
        H = H + beta*I
    end
    xn = x0 - H \ dot_f(x0)
    return xn
end
```

```julia
fig = figure(figsize = (4, 3))
xguess = 0.0
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
xnew = regularized_newton_step(xguess[end])
xguess = [xguess; xnew]
fig = figure(figsize = (4, 3))
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
function backtracking_regularized_newton_step(x0)
    beta = 1.0
    c = 0.5
    b = 0.1
    H = ddot_f(x0)
    while H <= 0  # isposdef https://docs.julialang.org/en/v1/stdlib/LinearAlgebra/#LinearAlgebra.isposdef
        H = H + beta*I
    end
    delta_x = - H \ dot_f(x0)
    alpha = 1.0
    while f(x0 + alpha * delta_x) > f(x0) + b * alpha * delta_x * dot_f(x0)
        alpha = c * alpha
    end
    xn = x0 + alpha * delta_x
    return xn
end
```

```julia
fig = figure(figsize = (4, 3))
xguess = 0.0
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
xnew = backtracking_regularized_newton_step(xguess[end])
xguess = [xguess; xnew]
fig = figure(figsize = (4, 3))
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```
